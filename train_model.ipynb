```python
# Replace unrealistic ages
users['Age'] = users['Age'].apply(lambda x: x if 5 < x < 100 else np.nan)

# Fill missing with median
users['Age'].fillna(users['Age'].median(), inplace=True)
```

```python
books['Year-Of-Publication'] = pd.to_numeric(books['Year-Of-Publication'], errors='coerce')
```

```python
books['Year-Of-Publication'] = books['Year-Of-Publication'].apply(
    lambda x: x if 1900 < x <= 2025 else np.nan
)
```

```python
books['Year-Of-Publication'].fillna(books['Year-Of-Publication'].median(), inplace=True)
```

```python
books['Publisher'].fillna('Unknown', inplace=True)
```

```python
ratings = ratings[ratings['Book-Rating'] != 0]
```

```python
print("Books:", books.shape)
print("Ratings:", ratings.shape)
print("Users:", users.shape)
```

```python
print(books['ISBN'].dtype)
print(ratings['ISBN'].dtype)
```

```python
books['ISBN'] = books['ISBN'].astype(str).str.strip().str.upper()
ratings['ISBN'] = ratings['ISBN'].astype(str).str.strip().str.upper()
```

```python
users['User-ID'] = users['User-ID'].astype(int)
ratings['User-ID'] = ratings['User-ID'].astype(int)
```

```python
common_isbn = set(books['ISBN']).intersection(set(ratings['ISBN']))
print("Common ISBNs:", len(common_isbn))
```

```python
df = ratings.merge(books, on='ISBN')
df = df.merge(users, on='User-ID')

print("After merge:", df.shape)
```

```python
print("Before filtering:", df.shape)
```

```python
df['User-ID'].value_counts().describe()
```

```python
df['Book-Title'].value_counts().describe()
```

```python
active_users=df.groupby('User-ID').count()['Book-Rating']>=50
active_users=active_users[active_users].index

df=df[df['User-ID'].isin(active_users)]
```

```python
popular_books = df.groupby('Book-Title').count()['Book-Rating'] >=20
popular_books = popular_books[popular_books].index

df = df[df['Book-Title'].isin(popular_books)]
```

```python
pt = df.pivot_table(index='Book-Title', columns='User-ID', values='Book-Rating')
pt.fillna(0, inplace=True)
```

```python
print(pt.shape)
```

```python
from sklearn.metrics.pairwise import cosine_similarity

similarity_scores = cosine_similarity(pt)
```

```python
def recommend(book_name):
    if book_name not in pt.index:
        return["book not found"]
    index=np.where(pt.index==book_name)[0][0]
    similarity_scores[index]
    similar_items=sorted(
        list(enumerate(similarity_scores[index])),
        key=lambda x:x[1],
        reverse=True
    )[1:6]
    
    data = []

    for i in similar_items:
        book_title = pt.index[i[0]]
        
        temp_df = books[books['Book-Title'] == book_title]
        
        data.append({
            "title": book_title,
            "author": temp_df['Book-Author'].values[0],
            "image": temp_df['Image-URL-M'].values[0]
        })

    return data
    
```

```python
import pickle

pickle.dump(pt, open('pt.pkl', 'wb'))
pickle.dump(similarity_scores, open('similarity.pkl', 'wb'))
pickle.dump(books, open('books.pkl', 'wb'))
```

```python
recommend("The Hobbit")
```
